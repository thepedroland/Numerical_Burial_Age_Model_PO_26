```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(19680801)
pts = np.random.rand(30)*.2
pts[[3, 27]] += 18 #put the 4th and 18th dots as outliners

fig, ax = plt.subplots()

ax.plot(pts)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, sharey=True, dpi=100)

ax1.hist(pts)
ax2.hist(pts)
ax1.set_xlim(0, 0.2)  #  set x axis range of subplot 1
ax2.set_xlim(0.5, 30)  # set x axis range of subplot 2

ax1.spines['right'].set_visible(False)
ax2.spines['left'].set_visible(False)
d = .85
#Draw markers and the broken places
kwargs = dict(marker=[(-1, -d), (1, d)], markersize=15,
              linestyle='none', color='r', mec='r', mew=1, clip_on=False)
ax1.plot([1, 1], [1, 0],transform=ax1.transAxes, **kwargs)
ax2.plot([0, 0], [0, 1], transform=ax2.transAxes, **kwargs)
plt.tight_layout()
plt.show()

```
